```python
import getpass, subprocess

username = input("Login GitHub: ").strip()
repo = "dwton_all"
token = getpass.getpass("GitHub token (nie pokaże się): ")
```

```python
url = f"https://{username}:{token}@github.com/{username}/{repo}.git"

# ustaw remote (bez tokenu w konfiguracji)
subprocess.run(["git", "remote", "remove", "origin"], capture_output=True)
subprocess.run(["git", "remote", "add", "origin",
                f"https://github.com/{username}/{repo}.git"])
subprocess.run(["git", "branch", "-M", "main"])

# ściągnij to, co już jest w repo (README itp.) i połącz historie
pull = subprocess.run(
    ["git", "pull", url, "main", "--allow-unrelated-histories", "--no-edit"],
    capture_output=True, text=True)
print("PULL:\n", pull.stdout, pull.stderr)

# wypchnij
push = subprocess.run(["git", "push", "-u", url, "main"],
                      capture_output=True, text=True)
print("PUSH:\n", push.stdout, push.stderr)

del token, url
print("Gotowe.")
```
