## 7 Poziomów Autonomii AI

W DataWorkshop używamy autorskiej mapy, która porządkuje chaos w głowie. Co to jest? Już tłumaczę.

Wyobraź to sobie jako suwak:
- Z jednej strony masz **Human-in-the-Loop (100%)** – człowiek musi wszystko sprawdzać.
- Z drugiej strony masz **Pełną autonomię (0%)** – maszyna decyduje sama (*Science Fiction*).

### Paradoks: Autonomia vs kontrola

Zanim przejdziemy dalej, musisz zrozumieć jedną fundamentalną zasadę, którą kierujemy się w DataWorkshop.
Często słyszę: *"Chcę dać AI autonomię, żeby nie musieć go pilnować"*.

**To błąd logiczny.**
Uznajemy, że jeśli ktoś zaczyna tworzyć bardziej autonomiczne systemy (tym bardziej na poziomie 4–5, Software 3.0), to musi obowiązywać inna zasada (której sami mocno się trzymamy i której uczymy innych):

> **Im wyższa Autonomia, tym wyższa musi być Kontrola.**

Gdy dajesz modelowi dostęp do narzędzi (API, bazy danych, wysyłka maili i wiele innych rzeczy), ryzyko rośnie wykładniczo. Dlatego musisz zwiększyć kontrolę. Ale uwaga – są dwa rodzaje kontroli:

1.  **Kontrola manualna (human-in-the-Loop):**
   - Człowiek siedzi i zatwierdza każdy krok (poziom 1).
   - To działa na początku, ale **nie skaluje się**. To wąskie gardło. To zaprzeczenie automatyzacji.
2.  **Kontrola systemowa (architektura):**
    - To przemyślana struktura, twarde ograniczenia, walidacja typów danych.
    - To system, który **uniemożliwia** pewne działania (np. "AI nie może wykonać przelewu powyżej 100 PLN bez dodatkowego klucza").

### Dlaczego o tym mówię teraz?
Ponieważ **Poziom 2 (Programowalne AI)**, którym zajmiemy się za chwilę, to fundament kontroli systemowej.
Nie da się kontrolować "lania wody" (czysty tekst). Ale da się doskonale kontrolować **strukturę**.

Jeśli nauczysz model zwracać przewidywalny schemat, zyskujesz władzę nad jego autonomią. Możesz automatycznie odrzucić błędne odpowiedzi, zanim trafią do użytkownika.

W DataWorkshop budujemy systemy autonomiczne (Software 3.0), ale **bezpieczne**. Nie poprzez sadzanie ludzi do pilnowania botów, ale poprzez projektowanie nieprzekraczalnych barier w kodzie. To jest ważne!

Zauważ kluczową zależność: **Im wyższy poziom autonomii, tym mniej człowieka w procesie, ale więcej człowieka w architekturze.**

Gdy system działa samodzielnie (autonomia), nie ma tam miejsca na "ręczne sterowanie". Dlatego musisz włożyć ogromny wysiłek **przed** uruchomieniem, projektując sztywne ramy kontroli.

Pamiętaj: **LLM jest z definicji niedeterministyczny** (nigdy nie masz 100% pewności, co zwróci).
Nie możesz budować systemów bezpieczeństwa opartych tylko na LLM-ach ("Proszę, nie rób tego"). To za mało.
Aby ustabilizować ten żywioł, musimy połączyć **Software 3.0** (kreatywny, ale chaotyczny model) z **Software 1.0** (stabilnym, nudnym kodem). Tylko to połączenie daje kontrolę.

#### 👉 Poziom 2: Programowalne AI (Software 3.0)
- **To jest cel tego warsztatu.**
- Model przestaje lać wodę. Zaczyna zwracać **strukturę** (Structure OUtput).
- Dzięki temu możesz podpiąć go pod kod w Pythonie. To moment, w którym AI staje się elementem systemu IT, a nie tylko zabawką.

Od tego poziomu już można wdrażac na produkcje. Co ciekawe, jak ślyszysz hasło Agent AI to możesz pomyśleć, że chodzi o któryś bardziej zaawansowany sposób. To nie prawda! To działa inaczej. Agenci AI, podobnie jak ludzi moga mieć rożne talenty, zdolności. Dlatego prawidłowa o tym myśleć, jak któs mówi, że wdrożyl Agent AI, to zapytaj z jakiego poziomu. Bo Agent i nawet wiele agentów również moga być na poziomie drugim. 

To samo dotyczy mity, że realna wartość dla biznesu zaczyna się dopiero do 5 poziomu. Nie! Tak naprawdę, realna wartość już jest na drugim poziomie, i wiele biznesów jeszzce tego nie dostrzegała. Owszem, im wyżej, tym więcej wartości można dostarczyć (innej złożności problemy rozwiazać), ale kluczowa myśl, że już od poziomu 2 można robić wdrożenia na rpoduckje i rozwiazywazc realne problemy.

ALE muszę tutaj ostrzec o jednej ważnej rzeczy, która dzieje się regularnie (między innymi to była motywacja do powstania takiej klasyfikacji). Większość projektów AI (czyli w tym przypadku takich, które są oparte na LLM) kończy się fiaskiem. Dlaczego? Powodów jest wiele, ale jednym z nich jest to, że biznes często oczekiwał (bo *hype* mocno rozgrzał oczekiwania) rozwiązania na poziomie 5 lub nawet 7, podczas gdy zespół ma kompetencje na poziomie 1. To nigdy się nie uda.

Dlatego bardzo ważne jest, aby przed rozpoczęciem projektu zrobić klasyfikację problemu: sprawdzić, czego oczekuje biznes i na jakim poziomie można to dostarczyć, realnie ocenić kompetencje zespołu, a wtedy będzie już widać, jaka jest szansa, że uda się to dowieźć na czas, w budżecie i — co najważniejsze — żeby działało zgodnie z oczekiwaniami lub nawet lepiej.

### Bielik 11B v3

W tym warsztacie będziemy używać model **Bielik v3** (30 grudnia 2025).

Co siedzi pod maską (zgodnie z [paperem](https://arxiv.org/pdf/2601.11579)):

*   **Baza:** Mistral 7B v0.2.
*   **Skalowanie:** Depth up-scaling do **11 miliardów parametrów** (11B).
*   **Trening:** 4-etapowy pipeline: Continuous Pre-training $\rightarrow$ SFT $\rightarrow$ DPO $\rightarrow$ Reinforcement Learning.

Wyniki? Bielik 11B v3 w testach przebija modele, które są **2 do 6 razy większe**.

Zanim jednak przejdziemy do kodu, musimy sobie wyjaśnić jedną rzecz z inżynierską szczerością. Często słyszy się hasło "Polski Model AI". Jako inżynierowie musimy być precyzyjni. Bielik nie był trenowany od zera. Już wyjaśniam co to oznacza w praktyce.

#### Jak powstają modele?
Proces trenowania LLM składa się z dwóch głównych faz:

1.  **Faza fundamentalna (Pre-training):** To tutaj model uczy się "rozumieć świat", logikę, fizykę i język. To kosztuje dziesiątki, setki milionów dolarów (a czasem nawet i miliardy) i wymaga potężnych klastrów obliczeniowych (między innymi OpenAI teraz inwestuje w to miliardy, a [Oracle](https://finance.yahoo.com/news/oracle-big-50-billion-bet-132458618.html) z kolei być może... dużo straci (chociaż może nie upadnie "tak łatwo" — bo jest za duży, ale kto wie)! To dłuższa historia, ale skracając ją, chodzi o gigantyczne inwestycje w klastry GPU). Na to stać tylko garstkę firm na świecie, takich jak OpenAI, Gemini, Anthropic, a jeśli mówimy o modelach otwartych — LLaMA, Mistral AI czy DeepSeek.
2.  **Faza specjalizacji (Fine-tuning/Post-training):** To etap, gdzie bierzemy "wyedukowany" model i uczymy go konkretnych zadań, języka czy stylu bycia.

Bielik **nie jest** trenowany od zera (od fazy 1). 

Bielik stoi na ramionach innego modelu. Jego fundamentem jest **Mistral** (np. dla Bielika v3 to Mistral 7B v0.2). Pewnie mógłby być też trenowany od zera — to można dyskutować — ale tak jak napisałem wcześniej, ktoś musiałby podjąć decyzję o zainwestowaniu naprawdę większych pieniędzy. Z punktu widzenia kraju być może nie są to aż tak duże kwoty ([na jachty](https://www.money.pl/gospodarka/kupili-dwa-jachty-za-dotacje-z-kpo-nie-dostalismy-tych-pieniedzy-na-wakacje-7187027702864640a.html) pewnie wydaje się więcej), jednak nikt takiej decyzji nie podjął. Zresztą, mówiąc o Bieliku, trzeba powiedzieć, że powstał on właściwie wyłącznie na początku dzięki entuzjazmowi (i to jest warto podziwu i dużego szacunku!), a później pojawiały się już różne próby pozyskania funduszy. Może warto dodać, dla kontekstu, że równoległa ścieżka, jeśli chodzi o [PLLuM](https://pllum.org.pl), wygląda inaczej — tu faktycznie pozyskał finansowanie, ale zwróć uwagę na kwotę: jedynie 10 mln zł, czyli około 3 mln dolarów (lub nawet mniej). Dlatego w tym przypadku była to również druga faza (oparta na modelu LLaMA, a nie na trenowaniu od zera modelu bazowego).

Wracamy do Bielika. Oznacza to, że DNA tego modelu, jego architektura i fundamentalne "rozumienie rzeczywistości" pochodzą od inżynierów z Francji (Mistral).

    - 👤 **`role: "user"`** – To Ty. Wpisujesz tu swoje polecenia ("Jak masz na imię?").
    - ⚙️ **`role: "system"`** – (Tego tu nie użyliśmy, ale użyjemy). To "głos z tyłu głowy" modelu. Tutaj wrzucamy kontekst biznesowy, dane z bazy wiedzy (RAG) albo instrukcje ("Jesteś ekspertem od Pythona"). Użytkownik końcowy tego nie widzi, ale to steruje zachowaniem modelu.
    - 🤖 **`role: "assistant"`** – To jest odpowiedź modelu. Kiedy LLM odpisuje, przyjmuje właśnie tę rolę.

3.  **Wyciągnięcie odpowiedzi**
    ```python
    response.choices[0].message.content
    ```
    Dlaczego to wygląda tak skomplikowanie?
    - `choices`: API może zwrócić kilka wariantów odpowiedzi naraz (np. 3 różne wersje maila). To jest lista
    - `[0]`: Nas interesuje zazwyczaj ten pierwszy (zwykle mamy jeden)
    -`message.content`: To jest to "mięso", czyli tekst, który wygenerował model


Model działa, przedstawia się i rozumie po polsku.
Ale zauważ jedną rzecz – zwrócił nam **luźny tekst**. Opowiedział o sobie.
Gdybyś chciał użyć tej odpowiedzi w kodzie (np. w `if` statement), miałbyś problem.

I to jest problem **Poziomu 1**.

### Eksperyment: Czy "lepszy prompt" wystarczy?

Możesz powiedzieć: *"Po co nam te klasy, biblioteki i kod? Czytałem na LinkedIn, że wystarczy dobrze napisać prompt i model zrobi to, co chcę."*

Sprawdźmy to. To klasyczny mit.
Zmieńmy prompt na bardziej "stanowczy". Dodajmy instrukcję: **"Zwróć tylko swoje imię!"**.

````python
response = client.chat.completions.create(
    model = "speakleash/Bielik-11B-v3.0-Instruct",
    messages = [
        {"role": "user", "content": "Jak masz na imię? Zwróć tylko swoje imię!"}
    ],
)

response.choices[0].message.content
````

### Co tu poszło nie tak?

1.  **Format:** Prosiliśmy o "tylko imię" (1-2 słowa), a mamy 2-3 zdania. Jeśli Twój kod spodziewał się krótkiego stringa, właśnie wyrzucił błąd.
2.  **Treść (ciekawostka):** Mimo że model nazywa się "Bielik" (tak jest w karcie modelu), jego trening (alignment) sprawił, że utożsamia się jako ogólny "Asystent" lub "AI".

To ważna lekcja: Model zwraca to, na co został wytrenowany, a niekoniecznie "fakt techniczny".
Ale mniejsza o imię. Na ten moment ważniejsze jest to, że **model zignorował polecenie dotyczące zwięzłości**.

Każdy ma prawo utożsamiać się z czym chce, ale w IT potrzebujemy **danych**, a nie uprzejmości.
Dlatego "zaklinanie promptem" (poziom 1) się nie skaluje.

Musimy przejść wyżej, czyli poziom 2.

### Poziom 2: Wymuszanie struktury (koniec z "laniem wody")

Załóżmy, że chcemy, aby napisał tylko swoje imię. 
W tym przypadku zdefiniujemy strukturę składającą się z jednego pola, niech to pole nazywa się dosłownie `"name"` (kod zawsze piszemy po angielsku) i będzie tekstowe.

To, co ma zrobić LLM, to uzupełnić tylko i wyłącznie to pole.

W poprzednim kroku model opowiedział nam o sobie całą historię. To fajne do czytania, ale fatalne do przetwarzania w kodzie.
Jeśli chcesz użyć imienia modelu w zmiennej (np. `current_model_name`), trzeba się pomęczyć np. ciąć ten tekst, szukać słów kluczowych... Koszmar.

Dlatego wkraczamy na **Poziom 2**. Zmusimy model, żeby przestał być poetą, a stał się **parserem**.

Zdefiniujmy prostą strukturę danych:

```python
class Response(BaseModel):
    name: str
```

#### Co tu się dzieje?

1.  **`class Response(BaseModel)`**:
    Używamy biblioteki **Pydantic**. To standard w Pythonie do walidacji danych.
    Definiujemy "formularz", który model musi wypełnić.

2.  **`name: str`**:
    To jest nasza instrukcja dla modelu. Mówimy mu:
    > *"Interesuje mnie tylko Twoje imię. I ma to być tekst (string). Nie chcę daty, nie chcę liczby, nie chcę opisu pogody. Tylko imię."*

### Dlaczego to jest krytyczne?
To jest moment, w którym LLM przestaje być "czarną skrzynką" wypluwającą słowa jak mu się chce.
Dzięki tej klasie tworzymy **kontrakt (API)** między Tobą a modelem.

- **Bez struktury (Poziom 1):** Dostajesz "Cześć! Jestem Bielik i lubię placki...".
- **Ze strukturą (Poziom 2):** Dostajesz obiekt `Response`, gdzie `response.name` to po prostu `"Bielik"`.

To różnica między chaosem a porządkiem. Teraz możesz bezpiecznie użyć tego wyniku w `if`, w bazie danych czy w API.

### `Structured Output`: Fundament Software 3.0

To, co teraz robimy, fachowo nazywa się **Structured Output**.
Warto rozróżnić dwa pojęcia, które często są mylone:

1.  **Function Calling:** To mechanizm, gdzie model decyduje, *którą funkcję wywołać* (np. `get_weather()`). To jest narzędzie do **działania**.
2.  **Structured Output:** To mechanizm, gdzie model ma po prostu **zwrócić dane** w konkretnym formacie (np. wyciągnąć imię z tekstu). To jest narzędzie do **parsowania**.

Choć "Structured Output" wyewoluował z "Function Calling", dzisiaj traktujemy go jako osobną, kluczową kategorię. 

W skrócie, dla Ciebie oznacza to, że warto poświęcić czas na poznanie idei Structured Output, ponieważ łączy w sobie wszystko, co potrafisz robić z function calling, ale w znacznie bardziej rozbudowany sposób. Nie chodzi tu tylko o wywołanie funkcji. Chodzi o utrzymanie struktury i pewności działania. Pamiętasz, jak mówiłem o autonomii — im większa autonomia, tym większa kontrola. *Structured Output* jest mostem, który umożliwia „dokręcanie śrubek” w tym obszarze.

#### Dwie drogi do struktury

Jak zmusić model, żeby zwrócił idealnego JSON-a zgodnego z naszą klasą Pydantic? Mamy dwie opcje:

1.  **Natywne wsparcie:**
    Giganci (OpenAI, Anthropic, Gemini, Google) wprowadzili to bezpośrednio do API. Wcześniej mówiłem na to - przyszłość, teraz już terażniejszość. Niestety, w szumie medialnym o "generowaniu obrazków" mało kto zauważył, że to **najważniejsza funkcja dla inżynierów**.
2.  **Biblioteka `Instructor` (iniwersalny adapter):**
    Co zrobić, gdy model (taki jak Bielik czy starsze LLaMA) nie ma natywnego wsparcia na poziomie API?
    Wtedy używamy biblioteki **`instructor`**.

### Dlaczego `Instructor` + Bielik?

Bielik (oparty na Mistralu) radzi sobie z JSON-em coraz lepiej, ale przy skomplikowanych strukturach potrafi się pogubić (zapomnieć klamry, pomylić typ pola). `Instructor` działa jak **strażnik**. Bierze odpowiedź modelu, sprawdza, czy pasuje do Twojej klasy Pydantic, a jeśli nie – prosi model o poprawkę (tzw. retry loop).

To sprawia, że nawet mniejszy model (11B) staje się użyteczny w systemach produkcyjnych.

## Użyjmy `instructor`

````python
client = instructor.patch(client, mode=instructor.Mode.MD_JSON)
````

### Co tu się stało?
Funkcja `patch` podmienia metody w kliencie OpenAI. Od teraz, zamiast zwracać surowy tekst, klient będzie wiedział, jak obsłużyć nasze klasy Pydantic.
Tryb `MD_JSON` (Markdown JSON) to wskazówka dla modelu: *"Jeśli masz problem, po prostu wypluj JSON w bloku kodu markdown, ja sobie to wyciągnę"*. To bardzo pomaga modelom open-source.

Mamy narzędzia. Mamy strukturę. Czas na test generalny. 👇

````python
class Response(BaseModel):
    name: str
````

````python
response = client.chat.completions.create(
    model = "speakleash/Bielik-11B-v3.0-Instruct",
    messages = [
        {"role": "user", "content": "Jak masz na imię?"}
    ],
    response_model=Response
)

response.name
````


### Co tu się właśnie wydarzyło? 

Uruchomiliśmy kod. Model zwrócił wynik.

Widzimy, że zwrócił imię. Jak już pokazałem wyżej, ciężko będzie wydobyć z Bielika, że ma na imię "Bielik", ponieważ utożsamia się bardziej z "AI", "asystentem" lub "AI Expertem". Teraz jest to już mniej istotne. Najważniejsze jest to, że nie zwrócił ściany tekstu, tylko wprost imię — inne, niż być może oczekiwaliśmy.


To, **co** dokładnie odpowiedział (treść), jest teraz mniej istotne (wiemy, że modele mają tendencję do skromności i czasem ukrywają swoje "imię" za ogólnikami).
Kluczowe jest to, **jak** to odpowiedział.

Nie masz ściany tekstu ("Cześć, jestem modelem językowym wytrenowanym przez... w czym mogę pomóc?").
Masz **obiekt**. Czyste pole `name`.

#### A Co zobaczył LLM? 

Kiedy dodaliśmy parametr `response_model=Response`, biblioteka `instructor` zrobiła za nas robotę.
Przetłumaczyła naszą klasę Pythonową na język zrozumiały dla maszyny – czyli **JSON Schema**.

Oprócz Twojego pytania (`"Jak masz na imię?"`), Bielik otrzymał "rozkaz" wypełnienia formularza. Zobaczył mniej więcej coś takiego:

```json
{
  "properties": {
    "name": {
      "title": "Name",
      "type": "string"
    }
  },
  "required": [
    "name"
  ],
  "title": "Response",
  "type": "object"
}
```

### Jak to interpretować?

1.  **`"type": "string"`**: Model wie, że musi wygenerować tekst, a nie liczbę czy tablicę
2.  **`"required": ["name"]`**: To jest bat na model. Musi wypełnić to pole. Nie może go pominąć
3.  **Kontekst nazw:** Zwróć uwagę, że nazwa klasy (`Response`) i nazwa pola (`name`) zostały przekazane do modelu

### Podsumowanie

Dobra robota. Pierwszy dzień za nami.

Może Ci się wydawać, że poświęciliśmy sporo czasu na kontekst i historię, ale w życiu nie ma drogi na skróty. Musieliśmy wylać ten fundament. Bez zrozumienia, gdzie jesteś na mapie 7 poziomów i dlaczego "zwykły czat" (Poziom 1) nie nadaje się do biznesu, budowałbyś zamki na piasku.

Co faktycznie mamy na koniec dnia?

Mamy środowisko. Mamy działającego Bielika, którego spięliśmy z biblioteką `instructor`. Ale najważniejsze jest to, że zmieniliśmy paradygmat. Przestajesz traktować model jako "kolegę do pogaduszek", a zaczynasz traktować go jak funkcję w kodzie, która ma przyjąć dane i zwrócić konkretny wynik (obiekt), a nie esej. Zaczynasz rozumieć, że JSON to Twoja jedyna tarcza przeciwko chaosowi i halucynacjom.

To był rozgrzewka. Jutro robimy kolejny krok. Będzie więcej wyzwań, ale na pewno sobie poradzisz!

Te ~4k "brudnych" ogłoszeń o pracę, które pokazałem w podglądzie, same się nie przeanalizują. Jutro rzucimy Bielika na głębszą wodę i zmotywujemy (zmusimy) go, by z tego bałaganu wyciągnął konkretną informację o poziomie stanowiska (Seniority). Zobaczysz, gdzie model się potyka i jak go "podnieść" kodem.
