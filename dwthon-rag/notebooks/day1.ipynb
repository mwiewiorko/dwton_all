## Software 1.0 → 2.0 → 3.0 (ale z innej strony)

Jeśli byłeś na DWthon #1 lub #2, ten koncept już widziałeś. Ale tym razem spójrzmy na niego z innego kąta — **kąta wiedzy**. Bo to właśnie ten kąt tłumaczy, dlaczego temat RAG wraca jak bumerang.

### Software 1.0: wiedza jest w kodzie

Klasyczne programowanie. Programista pisze reguły. Wiedza siedzi **w kodzie**:

```python
if cena > 1000 and klient.status == "VIP":
    rabat = 0.15
elif klient.status == "STANDARD":
    rabat = 0.05
```

To działa świetnie, kiedy:
- Reguły są jasne (prawo, procedury, logika biznesowa)
- Wiedza jest stabilna (nie zmienia się co tydzień)
- Da się to zapisać jako algorytm

Problem? Jeśli masz 10 000 faktur i każda wygląda trochę inaczej, nie napiszesz 10 000 reguł `if/else`. Próbowano. Nazywa się to "systemy ekspertowe" i to jest jedno z bardziej bolesnych wspomnień z historii AI.

### Software 2.0: wiedza jest w wagach modelu

Machine Learning. Model uczy się z danych. Wiedza siedzi **w wagach** — w miliardach parametrów, których nie da się łatwo zrozumieć ani edytować.

To działa świetnie, kiedy:
- Masz dużo danych treningowych
- Problem jest statystyczny (klasyfikacja, predykcja)
- Akceptujesz, że model czasem się myli

Problem? Nie wiesz, **dlaczego** model tak odpowiedział. Nie da się zmienić jednej reguły bez trenowania całości od nowa. To "czarna skrzynka".

### Software 3.0: wiedza jest w tekście

LLM rozumie język naturalny. Wiedza może siedzieć **w dokumentach, plikach, bazach wiedzy**. Mówisz do modelu: *"Mając ten kontekst, odpowiedz na to pytanie"*.

I tu dzieje się coś, czego **Software 1.0 i 2.0 nie potrafiły**:

- Wiedza jest **na zewnątrz** modelu
- Można ją **edytować** bez trenowania
- Można ją **audytować** (czytać, sprawdzać, wersjonować)
- Człowiek i maszyna **czytają ten sam artefakt**

To brzmi prosto. Ale konsekwencje są głębokie.

> **Software 3.0 to nie jest "LLM pisze za mnie kod".**
> **Software 3.0 to jest "wiedza staje się pierwszoklasowym obywatelem systemu — osobnym, zarządzanym, audytowalnym".**

I teraz — **jak tę wiedzę zorganizować**, żeby była naprawdę użyteczna? To jest cały temat RAG. I to jest cały temat tego warsztatu.

Karpathy swoim postem o wiki nie wymyślił nowej kategorii. On pokazał, że **kiedy LLM traktujesz jako "kompilatora wiedzy"** — zamiast tylko "generatora odpowiedzi" — możesz zbudować coś bardziej użytecznego. My w module 3 nazywamy to **Mapą Wiedzy**. Za chwilę do tego wrócimy.




## 7 Poziomów Autonomii AI — gdzie jest Poziom 3?

Kolejne odświeżenie dla wracających: mapa 7 Poziomów. Tym razem pokażę Ci ją z perspektywy **"ile wiedzy model ma do dyspozycji i skąd ją bierze"**.

<img src="../images/7_levels.jpg" width="900px" />

### 🟢 Poziom 1: Chat (wiedza z treningu)

Model odpowiada z tego, co zapamiętał w trakcie treningu. Nic więcej. Żadnego dostępu do Twoich danych, żadnej aktualizacji. Jeśli pytasz o coś, czego nie było w danych treningowych — dostaniesz halucynację (w najgorszym razie) albo uczciwe "nie wiem" (w najlepszym).

To jest ChatGPT bez pluginów, Claude bez projects, Gemini bez żadnego kontekstu.

### 🟢 Poziom 2: Structured Output (kontrakt z modelem)

Tutaj nie chodzi jeszcze o nową wiedzę. Tutaj chodzi o **kontrolę nad tym, CO model zwraca**. Zamiast "lania wody" — dostajesz strukturę: JSON, obiekt Pythona, tabelę.

To fundament, bez którego **nic dalej nie działa stabilnie**. Dlaczego? Bo Poziom 3 (RAG) to system, w którym wiele elementów musi sobie podawać dane. Jeśli na Poziomie 2 model nie potrafi stabilnie zwrócić struktury — to na Poziomie 3 cały pipeline się wywraca.

Pamiętaj:
- Jeśli byłeś na DWthon #1, szczegółowo budowałeś to z Bielikiem
- Jeśli nie byłeś — spokojnie, dzisiaj też będziemy tego używać

### 🟡 Poziom 3: RAG (zewnętrzna wiedza) ← **TUTAJ JESTEŚMY**

Model dostaje **dostęp do Twojej wiedzy**. Dokumentów, baz danych, plików, wiki.

Ale — i to jest kluczowe — **RAG to nie jest jedna technika**. To cała rodzina podejść:

| Wariant | Gdzie siedzi wiedza | Jak jest wyszukiwana |
|---|---|---|
| Naiwny RAG | Chunki w bazie wektorowej | Podobieństwo kosinusowe |
| BM25 / klasyczny search | Dokumenty + indeks odwrócony | Słowa kluczowe, TF-IDF |
| Structured retrieval | Baza danych / tabele | SQL, filtry po metadanych |
| Graph RAG | Graf encji i relacji | Nawigacja po grafie |
| **LLM Wiki** (Karpathy) | Pliki Markdown | LLM czyta indeks i wybiera strony |
| **Mapa Wiedzy** (nasza) | Ustrukturyzowane artefakty z metadanymi | Wieloetapowy retriever: filtry + LLM-based selection |

Widzisz, o co chodzi?

> **RAG to nie narzędzie. RAG to podejście: "model używa zewnętrznej wiedzy, a jakość systemu zależy od tego, JAK tę wiedzę przygotowałeś i JAK ją wyszukujesz".**

Baza wektorowa to **jedna z opcji**. Wiki Karpathy'ego to **inna opcja**. Nasze podejście z metadanymi tabel — to **trzecia opcja**. I każda z nich może być dobra lub zła w zależności od problemu.

### 🔴 Poziom 4-5: Pamięć, narzędzia, planowanie

Poziom 4 to agent z pamięcią i narzędziami. Poziom 5 to agent, który planuje adaptacyjnie.
**Ale uwaga:** Poziom 5 bez dobrego Poziomu 3 (wiedza) jest jak samochód bez paliwa. Agent musi skądś wiedzę brać. To właśnie RAG mu ją dostarcza.

Jeśli byłeś na DWthon #2 (OpenClaw), widziałeś na własne oczy, co się dzieje, gdy agent ma potężne narzędzia (Poziom 5), ale kiepski fundament wiedzy (Poziom 3 robiony po łebkach). Koszty lecą w kosmos, halucynacje kwitną, a efekty są... no cóż, efektowne na demo, ale beznadziejne w produkcji.

### Najważniejsza lekcja tej mapy

> **Nie przeskakujesz poziomów.**

Większość projektów RAG-owych kończy się fiaskiem, bo ktoś próbuje zbudować "inteligentnego agenta, który zna całą dokumentację firmową" (chce Poziom 4-5), ale na poziomie 3 zrobił naiwny chunking + Pinecone i nazwał to dniem. Potem się dziwi, że halucynuje.

Dzisiaj, przez te 3 dni, robimy Poziom 3 **uczciwie**. Krok po kroku. Z naciskiem na fundamenty, które przetrwają każdą nową modę.

### Paradoks Kontroli (przypomnienie)

> **Im wyższa autonomia, tym wyższa musi być kontrola.**

RAG daje modelowi dostęp do Twoich danych. Dostęp = władza. Władza = ryzyko. Im potężniejszy system budujesz, tym więcej pracy musisz włożyć w jego **kontrolę systemową** (walidacja, audyt, źródła).

To jest główny powód, dla którego w DataWorkshop uczymy robić RAG **audytowalny**. Każda odpowiedź ma wskazywać, z którego dokumentu, z której tabeli, z którego wiersza wyciągnięto informację. Bez tego — żadnego produkcyjnego wdrożenia.

## Czym RAG naprawdę jest (i czym nie jest)

OK, pora uderzyć w sedno. Wrócimy na chwilę do tego sporu "RAG umarł / RAG żyje", bo warto to domknąć raz, a dobrze.

### Mit #1: "RAG to baza wektorowa"

**Fałsz.** Baza wektorowa to jedno z narzędzi wyszukiwania. To tak, jak powiedzieć "samochód to silnik". Silnik jest ważny, ale bez reszty (podwozia, kierownicy, hamulców, kół) nie jedziesz nigdzie.

Kiedy Anthropic (Claude Code) zrezygnował z bazy wektorowej, nie zrezygnował z RAG. Dalej używa retrievalu — tylko opartego o pliki, grep, search po treści, i — najważniejsze — **LLM jako decydenta, co przeczytać**.

### Mit #2: "Karpathy pokazał coś rewolucyjnego"

**Też fałsz.** Karpathy pokazał bardzo konkretny i ładny wariant tego, a w praktyce inżynierskiej (m.in. w naszym module 3) — od co najmniej dwóch lat.

Jego główna teza brzmi:

> **"Nie wrzucaj surowych dokumentów do retrievera. Skompiluj je najpierw w uporządkowaną formę (wiki). Wtedy retrieval staje się proste, bo szukasz w dobrze zorganizowanej wiedzy."**

Czy to jest nowe? Nie. Czy jest potrzebne, żeby ktoś tak rozpoznawalny jak Karpathy to głośno powiedział, żeby rynek obudził się ze snu o "Vector DB rozwiąże wszystkie problemy"? **Tak. I bardzo dobrze, że to zrobił.**

### Co więc JEST RAG-iem tak naprawdę?

RAG to **system**, który składa się z co najmniej tych warstw:

```
┌─────────────────────────────────────────────┐
│  INGESTION (wczytanie i przygotowanie)      │
│  - parsing dokumentów                       │
│  - normalizacja formatów                    │
│  - ekstrakcja struktury (tabele, sekcje)    │
├─────────────────────────────────────────────┤
│  REPREZENTACJA WIEDZY (Mapa Wiedzy)         │
│  - ustrukturyzowane artefakty               │
│  - metadane generowane przez LLM            │
│  - powiązania między pojęciami              │
│  - audytowalne, edytowalne przez człowieka  │
├─────────────────────────────────────────────┤
│  RETRIEVAL (wyszukiwanie)                   │
│  - filtrowanie po metadanych                │
│  - LLM wybiera właściwe artefakty           │
│  - może używać BM25, embeddingów, grafu...  │
├─────────────────────────────────────────────┤
│  GENERATION (generowanie odpowiedzi)        │
│  - kontekst + pytanie → LLM                 │
│  - strukturalne odpowiedzi z cytatami       │
│  - confidence, sources, audyt               │
└─────────────────────────────────────────────┘
```

Widzisz, gdzie jest **prawdziwa robota**? W **reprezentacji wiedzy**. To jest to, co my w DataWorkshop nazywamy **Mapą Wiedzy**. I właśnie ją przez te 3 dni zbudujesz.

## Mapa Wiedzy - czym jest, czym nie jest

Zanim zaczniemy kod, domknijmy ten koncept. Bo to jest słowo-klucz całego warsztatu.

### Co to NIE jest

**Mapa Wiedzy to NIE jest prompt.** Prompt jest krótki, jednorazowy, używany w konkretnym zapytaniu. Mapa Wiedzy jest trwała, utrzymywana, rozwijana.

**Mapa Wiedzy to NIE jest baza wektorowa.** Bazy wektorowe to jedna z możliwych implementacji retrievalu. Mapa Wiedzy to warstwa **nad** tym — to sensownie zorganizowana wiedza.

**Mapa Wiedzy to NIE jest folder z PDF-ami.** Surowe dokumenty to "raw". Mapa Wiedzy to skompilowana, ustrukturyzowana, wzbogacona o metadane wersja tej wiedzy.

**Mapa Wiedzy to NIE jest graf wiedzy w klasycznym rozumieniu.** Graf wiedzy (knowledge graph) to mocno ustrukturyzowana forma (encje, relacje, trójki RDF). Mapa Wiedzy może być w różnych formach — pliki MD, tabele z metadanymi, graf, mix.

### Co to JEST

**Mapa Wiedzy to warstwa pośrednia** między surowymi danymi a modelem. To jej zadaniem jest:

1. **Reprezentować wiedzę w sposób dopasowany do typów pytań**, które naprawdę będą zadawane
2. **Być audytowalną** — ekspert biznesowy czyta plik .md albo kolumnę w tabeli i weryfikuje, czy informacja jest poprawna
3. **Być edytowalną** — można coś poprawić, dodać, usunąć bez "trenowania modelu"
4. **Zawierać bogate metadane** — tagi, kategorie, powiązania, confidence, źródła
5. **Być wersjonowana** — git, historia zmian, możliwość rollbacku
6. **Być używana zarówno przez człowieka, jak i przez LLM** — jeden artefakt, dwa odbiorców

Karpathy użył Markdownu + Obsidiana. My też tego użyjemy (bo to najprostszy, najbardziej uniwersalny format). Ale pamiętaj — **format jest drugorzędny**. Pierwszorzędna jest idea: **wiedza, którą kontrolujesz**.

````python
GITHUB_USER  = input("Twój GitHub username: ").strip()
GITHUB_TOKEN = getpass("Wklej Personal Access Token: ").strip()
REPO_NAME    = input("Nazwa repo (np. dwthon-rag-mapa-wiedzy): ").strip()
````

## Klonujemy Twoje prywatne repo do folderu

Jako wynik powinno pojawić się (lub Twoja nazwa folderu zamiast `dwthon-rag`):

```
✅ Repo sklonowane do ./dwthon-rag
```

````python
target_dir = f"./{REPO_NAME}"

if os.path.exists(target_dir):
    print(f"⚠️ Folder {target_dir} już istnieje. Pomijam klonowanie.")
else:
    clone_url = f"https://{GITHUB_USER}:{GITHUB_TOKEN}@github.com/{GITHUB_USER}/{REPO_NAME}.git"
    result = subprocess.run(
        ["git", "clone", clone_url, target_dir],
        capture_output=True, text=True
    )
    
    if result.returncode == 0:
        print(f"✅ Repo sklonowane do {target_dir}")
    else:
        print(f"❌ Coś poszło nie tak:")
        print(result.stderr)
````

### Sprawdźmy, co jest w sklonowanym repo

Jeśli widzisz `README.md` — wszystko działa. Ten folder (w chmurze, na naszym Jupyterze) jest teraz **podpięty do Twojego prywatnego repo na GitHub**. Jutro zaczniemy tu zapisywać pliki Mapy Wiedzy i wrzucać je na GitHub.

### Krok 5: Ustaw git config (żeby móc potem commitować)

````python
GITHUB_EMAIL = input("Twój email (ten z GitHub): ").strip()

subprocess.run(["git", "-C", target_dir, "config", "user.name", GITHUB_USER])
subprocess.run(["git", "-C", target_dir, "config", "user.email", GITHUB_EMAIL])

print(f"✅ Git config ustawiony dla {target_dir}")
````

### Krok 6: Mały test - zapiszemy plik i wrzucimy na GitHub
Tworzymy prosty plik testowy + `.gitignore` (na przyszłość) w sklonowanym repo.

Oczekiwany wynik:

### Krok 7: Commit i push na GitHub

Powinno być coś takiego:

```
[main 6319519] DWthon Day 1: hello world
 1 file changed, 4 insertions(+)
 create mode 100644 hello_dwthon.md
```

I dalej:

```
✅ Plik wypchnięty na GitHub!
   Sprawdź: ...
```

````python
os.chdir(target_dir)
subprocess.run(["git", "add", "hello_dwthon.md"])
subprocess.run(["git", "add", ".gitignore"])
subprocess.run(["git", "commit", "-m", "DWthon Day 1: hello world"])
push_result = subprocess.run(
    ["git", "push"],
    capture_output=True, text=True
)

if push_result.returncode == 0:
    print("✅ Plik wypchnięty na GitHub!")
    print(f"   Sprawdź: https://github.com/{GITHUB_USER}/{REPO_NAME}")
else:
    print("❌ Push nie wyszedł:")
    print(push_result.stderr)
````

Dlaczego akurat sprawozdania finansowe?

1. **Są pełne tabel**. A tabele to klasyczny przykład danych, które naiwny chunking niszczy doszczętnie.
2. **Są ustrukturyzowane, ale nie do końca**. Każda spółka ma swój format, swoje konwencje nazewnicze.
3. **Mają realną wartość biznesową**. Pytania typu "jaki był wynik netto spółki X w Q3 2024?" to nie akademickie ćwiczenie — to realne zapytania analityków.
4. **Są publicznie dostępne**. Więc nie gryziesz się z NDA, nie trzeba niczego anonimizować.
