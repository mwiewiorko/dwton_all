## Software 3.0, czyli koniec mitu "trenowania"

Zanim wejdziemy w pamięć, muszę rozprawić się z nieporozumieniem, które krąży w biznesie i myli nawet dobrych inżynierów.

Cofnijmy się do fundamentów - trzy paradygmaty budowania oprogramowania:

- **Software 1.0 (klasyczny kod):** człowiek pisze reguły. Logika siedzi w `if/else`. Chcesz zmienić zachowanie? Zmieniasz kod. Przewidywalne, ale nie ogarnie chaosu prawdziwego świata.
- **Software 2.0 (klasyczny ML):** nie piszesz reguł, tylko **trenujesz model na danych**. Wiedza siedzi w wagach. Tu słowo "trening" ma twarde, matematyczne znaczenie - aktualizujesz wagi (*backpropagation*).
- **Software 3.0 (LLM-y):** bierzesz **gotowy, potężny model**. On jest **zamrożony**. Jego wagi się nie zmieniają. Ani na jotę. I tu pojawia się problem. Rozmawiasz z klientem, a on mówi: *"Chcemy, żeby model uczył się w trakcie rozmowy"* albo *"System ma się dotrenować na naszych użytkownikach"*. Inżynier z krwi i kości dostaje wtedy gorączki: *"Przecież te modele się nie trenują! One są statyczne!"*. I klient, i inżynier mają rację jednocześnie. Trzeba tylko zmienić perspektywę.

### Przełamanie

Technicznie model LLM się nie zmienia. Ale **agenta AI trenujemy**. Tylko że ten "trening" to nie zmian "wag modelu". To **rzeźbienie pamięci**:

1. Mamy mechanizm, który selekcjonuje i magazynuje kluczowe informacje o użytkowniku i procesie.
2. Wyciągamy z pamięci właściwe kawałki we właściwym momencie.
3. Wstrzykujemy je dynamicznie do promptu.

I to jest cały trik:

> **Hiperpersonalizacja to iluzja trenowania, uzyskana dzięki perfekcyjnej inżynierii pamięci.**

Model "pod spodem" jest ten sam dla wszystkich. Ale dla konkretnego użytkownika zachowuje się unikalnie. Adaptuje się. Z punktu widzenia użytkownika - *"model się nauczył"*. Z punktu widzenia inżyniera - zbudowaliśmy architekturę **stateful**, która inteligentnie zarządza kontekstem.

*A fine-tuning? W ~99.99% przypadków, w których ludzie po niego sięgają, to armata na wróbla - droższa, wolniejsza i mniej elastyczna niż dobra inżynieria pamięci. Ale to temat na inny dzień.* Swoją drogą, nawet [OpenAI](https://developers.openai.com/api/docs/deprecations#update-to-openais-self-serve-fine-tuning) już to przyznało i po cichu wyłącza możliwość fine-tuningu po kilku latach. Zwróć uwagę, że OpenAI zrobiło fine-tuning banalnie prostym. Wrzuć dane. Kliknij. Masz swój model. A teraz po cichu to zwija:
- 7 maja 2026 – nowe organizacje już nie odpalą treningu.
- 2 lipca 2026 – koniec dla tych, którzy nie wykonywali inferencji przez 60 dni.
- 6 stycznia 2027 – nawet aktywni klienci nie utworzą nowego modelu.

Dlaczego? Bo w praktyce coraz częściej lepsze rezultaty dają dobrze napisane prompty, RAG i narzędzia niż utrzymywanie osobnych modeli. Fine-tuning nadal ma swoje zastosowania, ale dla większości przypadków przestał być domyślnym wyborem. To dość wyraźny sygnał, w którą stronę zmierza cały rynek.

**Ważny wniosek dla Ciebie na dziś:** w Software 3.0 nie walczysz o lepsze wagi. Walczysz o lepszą **pamięć i kontekst**. I to jest umiejętność, której nie kupisz w `pip install`.

## Gdzie na mapie jest pamięć? (7 Poziomów Autonomii AI)

Żeby nie zgubić się w hype'ie, używamy prostej mapy. Wyobraź sobie suwak autonomii - od "człowiek pilnuje każdego kroku" do "system działa sam".

**🟢 Poziom 1 - Chat.** 99% rynku. "Napisz mi maila". Model gada, ale nic nie "robi" systemowo. Zero powtarzalności.

**🟢 Poziom 2 - Programowalne AI.** Model przestaje lać wodę i zaczyna zwracać **strukturę** (JSON, obiekt). Dopiero tu AI wchodzi do systemu IT jako komponent. To fundament wszystkiego wyżej.

**🟡 Poziom 3 - RAG.** Model dostaje dostęp do **zewnętrznej wiedzy** (dokumenty, bazy). Pytasz o coś, on szuka i odpowiada.

**🔴 Poziom 4 - Pamięć i narzędzia.** Model zaczyna **pamiętać Ciebie** (kontekst, preferencje, historię, postęp) i **działać** (narzędzia). **Tu dzisiaj mieszkamy.**

**🔴 Poziom 5 - Planowanie.** Agent nie tylko ma narzędzia, ale planuje i adaptuje plan po drodze (np. Claude Code).

**Poziom 6-7** - meta-autonomia i AGI. Na razie sci-fi.

### Zapamiętaj to jedno rozróżnienie - bo wiele osób z internetu je myli:

> **RAG (poziom 3) ≠ Pamięć (poziom 4).**

- **RAG** to **encyklopedia**. Wyszukiwanie faktów o świecie. "Jaka jest polityka zwrotów?", "Co mówi dokumentacja?". Retrieval i tyle.
- **Pamięć** to **relacja**. Stan o *Tobie* i o *procesie*, który się **aktualizuje, konsoliduje, rozwiązuje konflikty i wygasza**. "Jesteś Markiem, uczysz się do prezentacji w NY, ostatnio kuł Cię Past Simple".

RAG odpowiada na pytanie *"co wiem o świecie?"*. Pamięć odpowiada na pytanie *"kim jesteś i gdzie jesteśmy w naszej wspólnej historii?"*.

Można mieć świetny RAG i zero pamięci. I odwrotnie. To osobne byty. Przez cały ten warsztat zajmujemy się **poziomem 4**.

*Ważne, żeby zapamiętać jedno: nie przeskakuje się poziomów. Nie zbudujesz agenta, który "planuje i pamięta" (poziom 5), jeśli na poziomie 4 pamięć robisz po łebkach. To najczęstszy powód, dla którego efektowne dema nie przeżywają zderzenia z produkcją.*

## Kontekst vs pamięć - dwa słowa, które trzeba rozdzielić

To najważniejsze rozróżnienie dnia. Zapamiętaj tak:

> **Kontekst decyduje, co model widzi TERAZ. Pamięć decyduje, co PRZETRWA.**

- **Kontekst** = okno kontekstowe. Ulotne. To, co wrzuciłeś do tego jednego wywołania. Znika po odpowiedzi.
- **Pamięć** = to, co zostaje między wywołaniami i między sesjami. Trwałe. Zarządzane.

I stąd bardzo praktyczny podział, który poprowadzi nas przez cały warsztat:

| | Pamięć **w-sesji** (short-term) | Pamięć **między-sesjami** (long-term) |
|---|---|---|
| Zakres | jedna rozmowa | wiele rozmów, dni, miesiące |
| Przykład | "przed chwilą ustaliliśmy datę" | "ten user zawsze woli zwięzłe odpowiedzi" |
| Problem | okno pęcznieje, model się gubi | jak przenieść wiedzę do następnej sesji |

**Sercem obietnicy tego DWthonu jest ta prawa kolumna** - agent, który **nie zaczyna od zera** przy każdym uruchomieniu. Ale żeby ją zrozumieć, najpierw musimy zobaczyć, jak pęka lewa.

### Dlaczego modele "gubią się" w rozmowie?

To nie jest opinia z LinkedIna. To wynik badań. Praca **"LLMs Get Lost in Multi-Turn Conversation"** (analiza 200 000+ symulowanych rozmów) pokazuje coś, co powinno przestraszyć każdego, kto buduje agentów:

- Wydajność spada średnio o **39%** w rozmowach wieloturowych względem pojedynczego, dobrze określonego pytania.
- Problem zaczyna się już **po 2 turach**.
- Mechanizm: model **wcześnie robi założenie**, przedwcześnie "strzela" rozwiązaniem, a potem **trzyma się swojego błędu**. Cytując autorów: *gdy LLM skręci w złą stronę, gubi się i nie potrafi wrócić.*

![](../images/llm_lost.jpg)
*[Źródło: LLMs Get Lost in Multi-Turn Conversation](https://arxiv.org/pdf/2505.06120)*

Wniosek? Sama "dłuższa rozmowa" to nie pamięć. To rosnący stóg siana, w którym model coraz trudniej znajduje igłę - a często sam dorzuca do stogu własne błędne domysły.

## Drabina naiwnych prób

Każdy, kto buduje agenta, wchodzi na tę samą drabinę. Każdy szczebel wygląda jak rozwiązanie. Każdy się łamie. Przejdźmy ją razem - bo na jej końcu czeka Twój blueprint.

### 🪜 Szczebel 1: brak pamięci (stateless)

Model LLM z natury jest **bezstanowy**. Dla niego każde zapytanie to "Dzień Świstaka" - budzi się, dostaje tekst, generuje odpowiedź i... traci wszystko. Wita Cię jak obcego.

Najbardziej ekstremalny znany medycynie przypadek to brytyjski dyrygent **Clive Wearing**. Po infekcji mózgu jego pamięć skróciła się do ~7-30 sekund. Clive żyje w wiecznym "teraz". Gdy żona wychodzi do kuchni i wraca po minucie, on wita ją z euforią i płacze ze szczęścia - bo jest przekonany, że nie widział jej od lat. Kilkadziesiąt razy dziennie. W dzienniku co chwilę zapisuje: *"8:31 - teraz naprawdę się obudziłem"*, skreślając poprzednie wpisy.

![](../images/clive_wearing.jpg)

## Wczytujemy biblioteki

from openai import OpenAI
from dotenv import load_dotenv

load_dotenv()

client = OpenAI(base_url="https://llm-api.dataworkshop.eu")

def ask(system_prompt, user_message, model="gpt-4o-mini"):
    response = client.chat.completions.create(
        model=model,
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": user_message}
        ],
        temperature=0.1
    )
    return response.choices[0].message.content

## Czujesz różnicę?

W pierwszym przypadku agent **reaguje na tekst**. Jest poprawny, ale "suchy" - to przesłuchanie. *"Jaki masz poziom? Co chcesz ćwiczyć? Jakie są Twoje cele?"*.

W drugim agent **adaptuje się do osoby**. *"Cześć, Marek! Super, że chcesz poćwiczyć. Jak poszło z przygotowaniem otwarcia prezentacji? Może chciałbyś je przedstawić, a ja pomogę Ci je poprawić? Potem możemy przejść do ćwiczenia, które pomoże Ci w płynnej prezentacji w Nowym Jorku. Co o tym myślisz?"*.

To ta sama różnica, co między recepcjonistą, który widzi Cię pierwszy raz, a starym znajomym, który zna Twój kontekst.

**Kluczowa obserwacja:** model jest **identyczny**. Ta sama temperatura, ten sam endpoint. Zmieniło się tylko jedno - **co podaliśmy mu w kontekście**. I to jest cała gra. Nie "lepszy model". **Lepsza pamięć.**

I tu rodzi się pokusa, która prowadzi nas na kolejne szczeble drabiny...


## 🪜 Szczeble 2–7: dlaczego naiwne rozwiązania się łamią

Skoro wstrzyknięcie kontekstu działa tak dobrze - to może wystarczy wrzucać coraz więcej? Zobaczmy, jak kolejne "oczywiste" pomysły rozbijają się o rzeczywistość.

**🪜 Szczebel 2 - "Dorzucę całą historię czatu".**
Najczęstszy błąd świata. *"Historia czatu = pamięć"*. Działa... do ~10-20 wiadomości. Potem: okno pęcznieje, koszt rośnie z każdą turą, model gubi wątek (patrz: te 39% spadku wydajności), a **nowa sesja = zero**. Zamykasz czat, otwierasz jutro - agent znów Cię nie zna. To bufor, nie pamięć.

![](../images/chat_history.jpg)

👉 **`Chat history to nie memory`.**

**🪜 Szczebel 3 - "To zwiększę okno kontekstu".**
*"Modele mają teraz milion tokenów, wrzucę wszystko!"*. Problem: "nieskończony kontekst" to mit. Płacisz za każdy token przy **każdym** zapytaniu (koszt), przeszukiwanie trwa (latencja), a im więcej wrzucisz, tym gorzej model wyciąga konkret (*Lost in the Middle*). Więcej ≠ lepiej. To dokładanie węgla do dziurawego pieca.
👉 **`Nie dokładaj tokenów. Zaprojektuj pamięć.`**

![](../images/window_context.jpg)

**🪜 Szczebel 4 - "To niech model streszcza rozmowę".** *(tu jest niespodzianka)*
Brzmi genialnie: streszczenie zachowa sens, obetnie tokeny. W praktyce **naiwna summaryzacja zawodzi** i to spektakularnie. Streszczenia **dryfują**, gubią to, co ważne, a jedna halucynacja z tury 3 staje się "faktem" w turze 8. Nawet Anthropic mówi wprost: *"compaction ≠ coherence"* - stratne streszczenia się rozjeżdżają. Problem nie leży w streszczaniu **per se** - leży w **niekontrolowanym, luźnym** streszczaniu. (Zapamiętaj to - jutro pokażę, jak robić to dobrze, czyli **strukturalnie**.)

**🪜 Szczebel 5 - "To niech zapisuje notatki do plików".**
Lepiej! Pliki są trwałe, człowiek może je przeczytać. Ale bez projektu plik rośnie w nieskończoność, nie ma struktury, agent nie wie, gdzie szukać, brak mechanizmu aktualizacji i wygaszania - a model potrafi nadpisać własne notatki. Plik to **nośnik**, nie architektura. Bez projektu masz cyfrowe wysypisko.

**🪜 Szczebel 6 - "To zrobię RAG / bazę wektorową".**
*"Wrzucę wszystko do bazy wektorowej i będzie pamięć!"*. Nie. RAG to **retrieval** - wyszukiwanie. Pamięć potrzebuje czegoś więcej: **aktualizacji, konsolidacji, rozwiązywania konfliktów, zapominania**. Ciekawostka: twórcy **Claude Code (Anthropic)** publicznie przyznali, że **wycofali się z bazy wektorowej**, bo inne podejście działało lepiej. Agent z samym RAG jest jak erudyta, który zna miliony faktów, ale nie pamięta ani jednej rozmowy z Tobą.
👉 **RAG (poziom 3) ≠ Pamięć (poziom 4).**

**🪜 Szczebel 7 - "To zainstaluję bibliotekę / memory-as-a-service".**
Ostatnia deska ratunku: *"ktoś to na pewno rozwiązał, `pip install`"*. Zła wiadomość: **nie ma standardu**. Cała branża - OpenAI, Anthropic, startupy - dłubie w tym po swojemu i nikt nie ma "tej jednej" biblioteki. Bo żadne narzędzie nie zna **Twojego** problemu biznesowego.
👉 **Nie ma `pip install memory_dla_mojego_agenta`. Musisz być architektem.**

## "Ale przecież modele będą coraz lepsze - po co mi to?"

Uczciwe pytanie, więc uczciwa odpowiedź. Sami inżynierowie z Anthropic mówią, że część problemów z kontekstem (tzw. *context rot*) to zjawisko **przejściowe** - kolejne modele radzą sobie lepiej, a niektóre protezy z dzisiejszych systemów jutro staną się zbędne. To prawda i warto to wiedzieć.

Ale nie daj się nabrać na wniosek *"to nie ma sensu inwestować w pamięć"*. Pamięć w naszym rozumieniu **nie jest łatką na słaby model**. Nawet gdyby jutro pojawił się model z nieskończonym, idealnym oknem, nadal chciałbyś zaprojektowanej pamięci - bo chodzi o **koszt** (nie wrzucasz całej historii przy każdym wywołaniu), **kontrolę** (decydujesz, co model widzi), **audyt** (wiesz, skąd wzięła się odpowiedź) i **stan biznesowy procesu** (gdzie jesteśmy w tej konkretnej sprawie). To są decyzje **architektoniczne**, nie modelowe. I one zostaną z nami niezależnie od tego, jak dobre będą modele.

## Zatrzymajmy się na chwilę: czym naprawdę jest "okno kontekstowe"?

Przeszliśmy przez całą drabinę. Ale żeby naprawdę zrozumieć, **dlaczego** te szczeble się łamią - zwłaszcza "dorzucę całą historię" i "zwiększę okno" - muszę dać Ci mentalny model, którego większość ludzi nie ma. (A jeśli dołączyłeś do nas pierwszy raz - spokojnie, wyjaśniam od zera.)

Szybkie przypomnienie: **okno kontekstowe** to wszystko, co model "widzi" w danym wywołaniu - instrukcje, historia rozmowy, dane, Twoje pytanie. Mierzy się je w **tokenach** (token to mniej więcej kawałek słowa).

I tu jest cały myk. Kiedy słyszysz "okno kontekstowe", prawdopodobnie myślisz o **jednej liczbie**: 128 tysięcy tokenów, milion tokenów. I tyle.

A ja patrzę na okno kontekstowe **trójwymiarowo**. To rozróżnienie zmienia wszystko.

### Wymiar 1: rozmiar techniczny

Ile tokenów model **fizycznie** przyjmie. To ta liczba z ulotki: 128k, 200k, milion. To jedyny wymiar, który zna większość ludzi - i to jest pułapka, bo prowadzi do myślenia: *"skoro technicznie zmieści się milion tokenów, to wrzucę tam wszystko i będzie pamięć"*.

### Wymiar 2: rozmiar efektywny

Tu robi się ciekawie. Mimo że **technicznie** zmieścisz milion tokenów, to **jakość** odpowiedzi zaczyna spadać dużo, dużo wcześniej. Często już po przekroczeniu 30-60 tysięcy (czasem 100k) dorzucanie kolejnych tokenów przestaje się opłacać - model gubi się we własnym stogu siana (pamiętasz *Lost in the Middle* i te 39% spadku w długich rozmowach?).

> Technicznie możesz. Efektywnie - nie warto.

### Wymiar 3: rozmiar semantyczny (tego prawie nikt nie widzi)

I to jest wymiar, który zmienia sposób myślenia. Bo tu **liczba tokenów w ogóle nie jest kluczowa**.

Możesz mieć naprawdę mało tokenów - powiedzmy 1000, 2000 - ale to, co w nich wrzucisz, wymaga **gigantycznego zaangażowania semantycznego**. Skomplikowany algorytm, w którym każdy element zależy od każdego. Coś rekurencyjnego. Zbiór sprzecznych ze sobą reguł.

Najprościej to poczuć po ludzku: **ile sekund potrzebujesz, żeby zrozumieć, co tam jest napisane?** Wyobraź sobie, że czytasz dziecku długą bajkę na dobranoc. Strony lecą, słowa płyną - i pewnie zauważyłeś, że czasem czytasz na głos, a po chwili nawet nie pamiętasz, co przeczytałeś. Dużo tokenów, **zero obciążenia semantycznego**. Spokojnie sobie płyniesz. A teraz wyobraź sobie jedną stronę skomplikowanego dowodu matematycznego albo zakręconego algorytmu, gdzie każda linijka zależy od poprzedniej. Ta sama liczba słów - ale musisz się **maksymalnie skupić**, wracać, trzymać wszystko w głowie naraz.

To jest właśnie okno semantyczne. Ono nie patrzy na to, **ile** wrzuciłeś, tylko na to, **czy już nie jest przeciążone**. I to nie teoria - to obserwacja, którą widać wprost w praktyce:

> **Im więcej reguł, warunków i instrukcji wrzucasz naraz, tym bardziej model głupieje.** Szczególnie te słabsze modele.

| # | Warstwa | Odpowiada na pytanie | Bez niej agent... |
|---|---|---|---|
| 1 | 🏛️ **Core** (Constitutional) | Kim jestem? Jakie mam nieprzekraczalne zasady? | jest nieprzewidywalny, "loose cannon" (Phineas Gage) |
| 2 | ⚡ **Working** | O czym rozmawiamy teraz? | gubi wątek "tu i teraz" (Clive Wearing) |
| 3 | 🎯 **Workflow** | Gdzie jestem w procesie? | zapętla się, nie dowozi celu, kończy w połowie |

To nie jest "optymalizacja" ani filozofia. Każda z tych warstw pełni konkretną funkcję, którą ewolucja (lub inżynieria) uznała za krytyczną. Usunięcie którejkolwiek to nie oszczędność - to zaproszenie do konkretnego rodzaju buga na produkcji.

*Ciekawostka: większość rozwiązań, o których głośno w branży, operuje na 4 "typach" pamięci (mniej więcej Working + Episodic + Semantic + Procedural). To dobry początek - ale to podzbiór. Prawdziwa robota zaczyna się, gdy dołożysz Core (bezpieczeństwo), Workflow (proces) i Resource (świat zewnętrzny) - i, co ważniejsze, gdy świadomie zdecydujesz, których w danym projekcie NIE brać.*

## Najważniejsza zasada: projektujesz, nie instalujesz

Zanim rzucisz się na budowanie wszystkich 7 warstw - stop. Głowna zasada, która oszczędzi Ci miesięcy (a może nawet i lata):

> **Nie istnieje idealna architektura pamięci.**

To jak z kodem: idealny kod to brak kodu (wtedy nie ma bugów). Idealna pamięć to brak pamięci (wtedy nie ma konfabulacji ani konfliktów). Ale wtedy system jest bezużyteczny (`stateless`). Sztuka polega na znalezieniu **minimum, które dowozi konkretny efekt**.

Dlatego nie zaczynamy od pytania "której bazy użyć?". Zaczynamy od **reverse engineeringu**:

1. Najpierw definiujemy **idealny scenariusz** - co musi się wydarzyć, żeby użytkownik pomyślał: *"Wow, ten system naprawdę mnie rozumie"*?
2. **Dopiero potem** cofamy się i projektujemy dokładnie te warstwy pamięci, które ten efekt umożliwiają. Reszta to szum.

Prosty agent (np. rezerwacja stolika w jednej knajpie) potrzebuje 3 warstw. Złożony tutor uczący języka miesiącami - więcej. Twoim zadaniem jako architekta jest **zdecydować**, a nie "wrzucić wszystko na wszelki wypadek".

## Podsumowanie Dnia 1

Dużo konceptów, mało kodu - i tak miało być. Pierwszy dzień to fundament, bez którego jutrzejszy kod byłby tylko klepaniem klas.

Co masz w głowie na koniec dnia:

1. **Software 3.0 to rzeźbienie pamięci, nie trenowanie wag.** Hiperpersonalizacja to iluzja treningu uzyskana dobrą inżynierią pamięci.
2. **RAG ≠ Pamięć.** Poziom 3 to encyklopedia, poziom 4 to relacja. Nie myl ich.
3. **Kontekst decyduje, co model widzi teraz; pamięć decyduje, co przetrwa.** A modele naprawdę gubią się w długich rozmowach (-39%).
4. **Drabina naiwnych prób łamie się na każdym szczeblu** - od historii czatu, przez "większe okno" i naiwne streszczanie, po "zainstaluję bibliotekę". Nie ma `pip install pamięć`.
5. **Pamięć to orkiestra 7 warstw**, a nie jedna szuflada. Każda warstwa to bezpiecznik przed konkretną awarią.
6. **Projektujesz, nie instalujesz.** Reverse engineering od efektu "wow, on mnie rozumie" do minimalnej architektury.
7. Masz **pierwszy blueprint pamięci** dla własnego agenta.

Poczułeś też na własne oczy tę różnicę - ten sam model, to samo pytanie, a dwa zupełnie różne światy. Cała gra rozegra się w tym, **co podamy modelowi w kontekście**.

## Co dalej? (Dzień 2)

Dziś narysowaliśmy architekturę "na papierze" i zrobiliśmy szybką symulację wstrzykiwania pamięci. Jutro **zanurzamy się w kod**.

Przełożymy blueprint na twarde struktury (Pydantic), zbudujemy **obiekt stanu agenta** (`AgentState`) jako jedyne źródło prawdy, dodamy logikę aktualizacji i poznasz mechanikę, która da Ci poczucie realnej władzy nad modelem - **fizyczne sterowanie tym, co agent w danej chwili widzi**. Zobaczysz też, jak robić kompresję pamięci **dobrze** (strukturalnie), skoro dziś ustaliliśmy, że naiwne streszczanie zawodzi.

Mniej literek, więcej kodu. Przygotuj kawę. ☕
